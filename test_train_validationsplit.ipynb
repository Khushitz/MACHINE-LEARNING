```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import r2_score
```

```python
data = pd.DataFrame({
    'Area': [800, 1200, 1500, 2000, 2200, 2500, 3000, 3500],
    'Location': ['Rural', 'Urban', 'Urban', 'SemiUrban', 'Urban', 'Rural', 'SemiUrban', 'Urban'],
    'Price': [2000000, 5000000, 6500000, 7000000, 9000000, 3000000, 7500000, 12000000]
})
print("Dataset:\n", data)
```

```python
#one hot encoding
encoder = OneHotEncoder(drop='first',sparse_output=False)
location_encoded = encoder.fit_transform(data[['Location']])
encoded_df=pd.DataFrame(location_encoded,columns=encoder.get_feature_names_out())
```

```python
#combine features
x=pd.concat([data[['Area']],encoded_df], axis=1)
y=data['Price']
```

```python
#train-test split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=42)
```

The test_size=0.25 argument specifies that 25% of the data will be allocated for the testing set, leaving 75% for the training set.
The random_state=42 argument ensures that the data is shuffled and split in the exact same way every time the code runs, which is crucial for reproducibility of results.


```python
#train model
model =LinearRegression()
model.fit(x_train,y_train)
```

```python
#evaluate
y_pred = model.predict(x_test)
#This step is crucial for evaluating how well the trained model performs on data it has not seen during training.

print("\nR2 Score:", r2_score(y_test,y_pred))
print("Intercept:",model.intercept_)
print("Coefficients:",model.coef_)
```
