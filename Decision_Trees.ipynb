```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
```

```python
df = pd.read_csv("/content/insurance_data.csv")
df
```

```python
x = df[['age']]
y = df['bought_insurance']
```

```python
#split data
x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2, random_state = 42
)
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
print("Accuracy:", model.score(x_test, y_test))
```

```python
plt.figure(figsize=(8,6))
tree.plot_tree(model, feature_names=['age'],class_names=['No','Yes'], filled = True)
plt.show()
```

LIKELY TO BUY INSURANCE

Age ≤ 24 → No

Age between 25–33 → Mostly No

Age > 33 → Mostly Yes

Age > 50 → Definitely Yes

It created rules automatically from data.

A Decision Tree is a supervised learning algorithm that:

Splits data into branches

Creates if-else rules

Used for classification and regression

Gini tells how mixed the data is.
Gini = 0 → Perfect (pure node)

Gini near 0.5 → Mixed data

Example:

[5,0] → Gini = 0 (pure)

[3,3] → Gini = 0.5 (very mixed)

Tree tries to reduce Gini at every split.

How Tree Decides Where to Split?

For classification:
**bold text**
Uses Gini or Entropy

For regression:
**bold text**
Uses MSE (Mean Squared Error)

Goal:

👉 Choose split that makes data most pure.

| Logistic Regression | Decision Tree       |
| ------------------- | ------------------- |
| Makes smooth curve  | Makes sharp splits  |
| Linear boundary     | Non-linear boundary |
| Needs scaling       | No scaling needed   |
| Less overfitting    | Can overfit easily  |


| Gini               | Entropy               |
| ------------------ | --------------------- |
| Faster to compute  | Slightly slower       |
| Default in sklearn | Used in ID3 algorithm |
| Similar results    | Similar results       |
