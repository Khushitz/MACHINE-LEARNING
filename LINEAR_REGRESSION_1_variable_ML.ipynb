```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model #library usefull for ml
```

```python
df = pd.read_csv("/content/homeprices.csv")
df
```

```python
plt.scatter(df.area, df.price, color='red',marker="+")
plt.xlabel('area')
plt.ylabel('price')

# suitable for linear regression
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
#  The code creates an instance of the LinearRegression model and fits it to data from a pandas DataFrame, specifically using the 'area' column as the feature and 'price' column as the target variable.

#Features are the input variables used to predict an outcome
#Target variable is the output, label, or dependent variable the machine learning model aims to predict.
```

```python
#predict PRICE of home who's AREA is 3300
reg.predict([[3300]])
```

```python
reg.coef_
# gives value of coeff based on y = mx +c
```

```python
reg.intercept_
# gives value of intercept based on y = mx +c
```
