**A dimensionality reduction technique.**

It reduces features while keeping maximum information.

🧠 **Intuition**

If you have:

100 features

Many are correlated

PCA reduces them to maybe:

10 important features

When to Use
**bold text**

Speed up training

```python
from sklearn.decomposition import PCA
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris

data = load_iris()

X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42
)


pca = PCA(n_components=2)
X_reduced = pca.fit_transform(data.data)

print(X_reduced.shape)
```

from sklearn.decomposition import PCA:
Imports the necessary class from the scikit-learn library.

pca = PCA(n_components=2):
Initializes a PCA object to keep only 2 principal components.

X_reduced = pca.fit_transform(data.data): Applies the PCA transformation to the data.
print(X_reduced.shape):

Prints the shape of the resulting reduced dataset
