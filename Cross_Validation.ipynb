**What is Cross Validation?**

Cross Validation is a technique used to:

Check how well your model will perform on unseen data.

Instead of splitting data only once into train & test…

We test the model multiple times on different splits.

This gives a more reliable performance estimate.

SOMETIMES TRAIN_TEST_SPLIT METHOD DOES NOT WORKS WELL

LIKE WHEN THE DATA IS SPLIT INTO LET'S SAY 0.3 TESTING SO IF A CHILD HAS PREPARED 70% FOR ALGEBRA AND THE REST 30% CAME FROM CALCULUS SO THAT WOULD NOT GIVE ACCURATE RESULTS HENCE THAT'S WHY SOMETIMES THIS BECOMES INACURATE.


**K-FOLD VALIDATION**
if there are dataset of 100 samples fold it into 5 samples and then perform steps like 1 fold use for test other 4 for training note down the scores then do the same but for 2nd fold note down the scores again and keep on doing this then later take down the averages

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
import numpy as np
from sklearn.datasets import load_digits
digits = load_digits()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(digits.data,digits.target,test_size =0.3)
```

```python
lr = LogisticRegression()
lr.fit(x_train, y_train)
lr.score(x_test, y_test)
```

```python
svm = SVC()
svm.fit(x_train,y_train)
svm.score(x_test,y_test)
```

```python
rf = RandomForestClassifier(n_estimators=40)
rf.fit(x_train,y_train)
rf.score(x_test,y_test)
```

```python
from sklearn.model_selection import KFold

data = [1,2,3,4,5,6,7,8,9]

kf = KFold(n_splits=3, shuffle=True, random_state=42)

for train_index, test_index in kf.split(data):
    print("Train indices:", train_index)
    print("Test indices:", test_index)
    print()
```

```python
def get_score(model, x_train,x_test,y_train,y_test):
  model.fit(x_train,y_train)
  return model.score(x_test,y_test)
```

```python
get_score(LogisticRegression(),x_train,x_test,y_train,y_test)
```

```python
get_score(SVC(),x_train,x_test,y_train,y_test)
```

```python
from sklearn.model_selection import StratifiedKFold
folds = StratifiedKFold(n_splits=3)
```

```python
scores_l = []
scores_svm = []
scores_rf = []

for train_index, test_index in kf.split(digits.data):

    x_train = digits.data[train_index]
    x_test  = digits.data[test_index]
    y_train = digits.target[train_index]
    y_test  = digits.target[test_index]

    scores_l.append(get_score(LogisticRegression(), x_train, x_test, y_train, y_test))
    scores_svm.append(get_score(SVC(), x_train, x_test, y_train, y_test))
    scores_rf.append(get_score(RandomForestClassifier(n_estimators=40), x_train, x_test, y_train, y_test))
```
