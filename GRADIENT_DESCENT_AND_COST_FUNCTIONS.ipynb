```python
# Gradient Descent = Going down in the direction where the error decreases fastest.
#like going downhill to reach the lowest point
#Gradient = Direction of the steepest slope
#Descent = Going downward
#basically aim to find value of x s.t y is minima
```

```python
# We’ll minimize the cost function:
#(m, b) = 1/n * sigma(y - (mx + b)) ^ 2
# ALSO KNOW AS MEAN SQ ERROR
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# sample data
x = np.array([1,2,3,4,5])
y = np.array([5,7,9,11,13])    # y = 2x +3


```

```python
# initialize params m is slope and b is intercept here
m = 0
b= 0
learning_rate = 0.01 # how big steps to take
iterations = 1000# how many times to update
n = len(x) # n = no of data points

#gradient descent
for i in range(iterations): # Improve the line 1000 times
  y_pred = m*x +b # This calculates predicted y using current m and b.AT FIRST ALL PREDICTIONS ARE 0 i.e m = b = 0

  #compute gradients # 👉 It measures how much we need to adjust m and b.

  dm= (-2/n) * np.sum(x*(y-y_pred)) # How wrong the slope is → dm

  db = (-2/n) * np.sum(y-y_pred) # How wrong the intercept is → db

  #update params
  # This is the improvement step. It slightly adjusts the line to reduce error
  m = m -learning_rate*dm

  b = b -learning_rate *db


print("Slope (m):",m)

print("Intercept (b):",b)
# After 1000 updates, it prints something close to:
# Slope ≈ 2
# Intercept ≈ 3
# BASICALLY PREDICTS SLOPE AND INTERCEPT

#Plot
plt.scatter(x,y)
plt.plot(x, m*x+b, color = 'red')
plt.show()

#Blue dots → original data
#Red line → predicted line
# If gradient descent worked correctly, the red line fits the blue dots perfectly.


```
