```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
#We are teaching the computer to predict Salary based on:

#Experience City
```

```python
data = pd.DataFrame({
    'City': ['Delhi', 'Mumbai', 'Chennai', 'Delhi', 'Mumbai', 'Chennai'],
    'Experience': [1, 3, 5, 2, 4, 6],
    'Salary': [30000, 50000, 70000, 35000, 60000, 80000]})
```

```python
print("Original Data:\n")
print(data)
```

```python
#onehot encoding
#Machine learning cannot understand text like "Delhi" or "Mumbai".
#So we convert cities into 0s and 1s.
#One-Hot Encoding → Convert categories into 0 and 1 columns.
#Dummy Variable → Same as one-hot but drop one column to avoid multicollinearity


encoder = OneHotEncoder(drop='first',sparse_output = False)
city_encoded = encoder.fit_transform(data[['City']])


# convert encoded array to df
encoded_df = pd.DataFrame(city_encoded, columns=encoder.get_feature_names_out())
```

```python
#combine numerical+encoded features
x = pd.concat([data[['Experience']], encoded_df], axis = 1)
y = data['Salary']

#We separate:

#X (inputs) → Experience + Encoded City

#y (output) → Salary

#So the model learns:
#Salary depends on Experience and City.
```

```python
# train linear regression
model = LinearRegression()
model.fit(x,y)

# salary = b +b1(exp)+b2(mumbai)+b3(delhi) since chennai drop = first dropped
```

```python
print(model.intercept_)
```

```python
print(model.coef_)
```
